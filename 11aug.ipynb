```python
# simple code for the follwing . 
# Bitwise operators demo

a = 10   # binary: 1010
b = 4    # binary: 0100

print("a =", a, "  (binary:", bin(a) + ")")
print("b =", b, "  (binary:", bin(b) + ")")
print()
```

```python
# Bitwise AND
and_result = a & b
print("a & b  =", and_result, "  (binary:", bin(and_result) + ")")

```

```python

# Bitwise OR
or_result = a | b
print("a | b  =", or_result, "  (binary:", bin(or_result) + ")")
```

```python
# Bitwise XOR
xor_result = a ^ b
print("a ^ b  =", xor_result, "  (binary:", bin(xor_result) + ")")

```

```python

# Left shift
left_shift = a << 1
print("a << 1 =", left_shift, "  (binary:", bin(left_shift) + ")")
```

```python
# Right shift
right_shift = a >> 1
print("a >> 1 =", right_shift, "  (binary:", bin(right_shift) + ")")
```
