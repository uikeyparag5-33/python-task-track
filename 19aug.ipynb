following are the method for list ..

append()
Adds an item at the end
insert()
Adds an item at a specific position
remove()
Removes a specific item
pop()
Removes an item using its position
sort()
Arranges items in order

reverse()
Reverse the list
a.reverse()

clear()
Remove everything
a.clear()

extend()
Adds multiple items from another list.

index()
Find position
a.index(5)



```python
cityname  = [" puna ", "   mumbai " , " delhi ", " kolkata ", " Bengaluru    ", " Hyderabad"  ] 
print(cityname ) 


cityname.append("Chennai")
print(" new using append  : ", cityname)


cityname.insert(3,"pune")                       # using the index position 
print(" new using insert   : ", cityname)


cityname.remove("pune")
print(" new using remove   : ", cityname)






cityname.pop (   1     )                    # using the index position 
print(" new using pop    : ", cityname)






cityname.sort(           )                     # using the first letter for sorting 
print(" new using sort : ", cityname )

# Python's sort() sorts strings alphabetically based on their characters, and spaces come before letters in the character ordering.

# So the leading spaces in ' puna ', ' delhi ', etc. affect the result.


newcity=[ " patna ", " surat ", "nagpur ","jaipur", " lucknow ",  "kanpur    "]

cityname.extend(newcity)
print(" using the extend : " ,cityname) 

print(cityname.index("nagpur "))
print(cityname )













```
